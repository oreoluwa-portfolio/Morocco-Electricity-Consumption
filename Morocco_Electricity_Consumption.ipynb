```
import pandas as pd
import matplotlib.pyplot as plt

# Ensure the data and variables are loaded inside this cell
df = pd.read_csv('powerconsumption.csv')
consumption_cols = ['PowerConsumption_Zone1', 'PowerConsumption_Zone2', 'PowerConsumption_Zone3']

# Calculate average consumption
zone_average = df[consumption_cols].mean()

# Plot the average consumption
zone_average.plot(
    kind='bar',
    figsize=(10,6)
)
plt.title('Average Electricity Consumption by Zone')
plt.xlabel('Zone')
plt.ylabel('Average Consumption Reading')
plt.xticks(rotation=0)
plt.show()
```

```
import pandas as pd
import matplotlib.pyplot as plt

# Ensure Datetime and Hour exist in the DataFrame
df['Datetime'] = pd.to_datetime(df['Datetime'], errors='coerce')
df['Hour'] = df['Datetime'].dt.hour

# Calculate the hourly consumption averages
hourly_consumption = df.groupby('Hour')[consumption_cols].mean()

# Plot the hourly pattern
hourly_consumption.plot(
    figsize=(12,6)
)
plt.title('Average Electricity Consumption by Hour')
plt.xlabel('Hour of the Day')
plt.ylabel('Average Consumption Reading')
plt.show()
```

```
import pandas as pd
import matplotlib.pyplot as plt

# Re-parse datetime and extract Month to ensure it exists
df['Datetime'] = pd.to_datetime(df['Datetime'], errors='coerce')
df['Month'] = df['Datetime'].dt.month

# Ensure monthly_consumption is calculated in this cell
monthly_consumption = df.groupby('Month')[consumption_cols].mean()

# Plot the monthly consumption pattern
monthly_consumption.plot(
    kind='line',
    marker='o',
    figsize=(12,6)
)
plt.title('Monthly Electricity Consumption Pattern')
plt.xlabel('Month')
plt.ylabel('Average Consumption Reading')
plt.show()
```

```
import pandas as pd
import matplotlib.pyplot as plt

# Ensure Datetime and DayofWeek exist in df
df['Datetime'] = pd.to_datetime(df['Datetime'], errors='coerce')
df['DayofWeek'] = df['Datetime'].dt.day_name()

# Calculate weekly consumption with proper day order
weekly_consumption = df.groupby('DayofWeek')[consumption_cols].mean()
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekly_consumption = weekly_consumption.reindex(day_order)

# Plot the weekly consumption pattern
weekly_consumption.plot(
    kind='bar',
    figsize=(12,6)
)
plt.title('Weekly Electricity Consumption Pattern')
plt.xlabel('Day of the Week')
plt.ylabel('Average Consumption Reading')
plt.xticks(rotation=45)
plt.show()
```

```
import seaborn as sns
import matplotlib.pyplot as plt

# Define variables locally to ensure self-containment
enviroment_cols = ['Temperature', 'Humidity', 'WindSpeed']
consumption_cols = ['PowerConsumption_Zone1', 'PowerConsumption_Zone2', 'PowerConsumption_Zone3']

# Calculate correlation matrix
correlation_data = df[enviroment_cols + consumption_cols]
correlation_matrix = correlation_data.corr(method='pearson')

# Generate heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5,
    fmt='.2f'
)
plt.title('Correlation Between Consumption and Environment Factors')
plt.show()
```
